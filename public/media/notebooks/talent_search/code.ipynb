```python
def is_digits_decreasing_or_equal(num):
    """
    Function to check if the digits of a number are in decreasing or equal order.
    """
    digits = list(str(num))
    for i in range(1, len(digits)):
        if digits[i] > digits[i - 1]:
            return False
    return True

# Find four-digit multiples of 81
for num in range(1000, 10000):
    if num % 81 == 0:
        if is_digits_decreasing_or_equal(num):
            print(num)

# Output
# This will print all four-digit multiples of 81 whose digits are in decreasing or equal order.

```
